### random model generate a sales data set




```python

import numpy as np
import pandas as pd

# Number of rows you want
num_rows = 1000

# Random product names
products = np.array(["Laptop", "Mobile", "Tablet", "Headphones", "Camera", "Monitor"])

# Random regions
regions = np.array(["Dubai", "Abu Dhabi", "Sharjah", "Ajman", "RAK"])

# Generate data
np.random.seed(42)   # for repeatable results

data = {
    "Sale_ID": np.arange(1, num_rows + 1),
    "Product": np.random.choice(products, num_rows),
    "Region": np.random.choice(regions, num_rows),
    "Quantity": np.random.randint(1, 10, num_rows),
    "Unit_Price": np.random.uniform(100, 2000, num_rows).round(2),
    "Sale_Date": pd.date_range("2024-01-01", periods=num_rows, freq="D")
}

# Create DataFrame
df = pd.DataFrame(data)

# Total sale value
df["Total_Sale"] = df["Quantity"] * df["Unit_Price"]

df.head()
```

```python
import numpy as np
from  datetime import datetime,timedelta
import random
```

```python
def sales_data_generator(no_samples=100):
  np.random.seed(56)
  product_id=np.random.randint(1,101,no_samples)
  quantity=np.random.randint(1,11,no_samples)
  price=np.random.uniform(500.4,999.4,no_samples).round(2)
  total_sales=(price*quantity).round(2)
  start_date=datetime(2025,1,1)
  dates=[(start_date + timedelta(days=i)).strftime('%Y-%m-%d')for i in range(no_samples)]
  data=[
          (
              dates[i],
              product_id[i].item(),
              quantity[i].item(),
              price[i].item(),
              total_sales[i].item()
          )
          for i in range(no_samples)
  ]
  return data
sales_data=sales_data_generator()
print("---SAMPLE DATA GENERATOR-----")

for x in sales_data[:]:
  print(x)
```
